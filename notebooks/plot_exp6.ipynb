```julia
using CSV
using DataFrames
using Plots
```

```julia
dir = "../data/results/exp6_avg_length"
df = CSV.read(joinpath(dir, "exp6_avg_length.csv"), DataFrame)
T = df.Actual_Avg_Len

p = plot(T, df.Base_Time_s, label = "Basic implementation", marker = :square, lw = 2)
plot!(p, T, df.Opt_Time_s, label = "Optimized implementation", marker = :diamond, lw = 2)
all(isnan, df.SPMF_Time_s) || plot!(p, T, df.SPMF_Time_s, label = "SPMF", marker = :circle, lw = 2)
plot!(p, T, df.Base_Build_s, label = "Basic: FP-tree construction only", marker = :utriangle, ls = :dot)
plot!(p, title = "Runtime vs. Avg. Length (minsup = $(df.Minsup[1]))",
      xlabel = "Average transaction length", ylabel = "Runtime (Seconds)", legend = :topleft, grid = true)
display(p)
savefig(p, joinpath(dir, "plot_runtime_vs_length.png"))
```

```julia
# Số itemset phổ biến và số node của FP-tree theo độ dài giao dịch
q = plot(T, df.Itemsets, label = "Frequent itemsets", marker = :circle, lw = 2, yscale = :log10)
plot!(q, T, df.Tree_Nodes, label = "FP-tree nodes", marker = :square, lw = 2)
plot!(q, title = "Itemsets and tree size vs. Avg. Length", xlabel = "Average transaction length",
      ylabel = "Count (log scale)", legend = :bottomright, grid = true)
display(q)
savefig(q, joinpath(dir, "plot_count_vs_length.png"))

# Thời gian theo số itemset sinh ra: thời gian khai thác tỉ lệ với kích thước output
r = plot(df.Itemsets, df.Base_Time_s, label = "Basic", marker = :square, lw = 2)
plot!(r, df.Itemsets, df.Opt_Time_s, label = "Optimized", marker = :diamond, lw = 2)
plot!(r, title = "Runtime vs. Number of Frequent Itemsets", xlabel = "Frequent itemsets",
      ylabel = "Runtime (Seconds)", legend = :topleft, grid = true)
display(r)
savefig(r, joinpath(dir, "plot_runtime_vs_itemsets.png"))

df
```
